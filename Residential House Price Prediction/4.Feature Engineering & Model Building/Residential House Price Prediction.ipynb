```python
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings("ignore")
import importlib
import Regression_utils
importlib.reload(Regression_utils)
from Regression_utils import *
```

```python
dataset=pd.read_csv("house_price_cleaned.csv")
dataset
```

```python
dataset.shape
```

```python
dataset.isnull().sum()
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
indep_X=dataset.drop("SalePrice", axis=1)
indep_X
```

```python
dep_Y=dataset["SalePrice"]
dep_Y
```

```python
# SelectK
summary_k = compare_k_values(
    indep_X,
    dep_Y,
    [35,20,25]
)

print("\nSELECT K")
print(summary_k)


# PCA
summary_pca = compare_pca_values(
    indep_X,
    dep_Y,
    [30,20,25]
)

print("\nPCA")
print(summary_pca)
```

```python
# RFE
summary_rfe = compare_rfe_values(
    indep_X,
    dep_Y,
    [15,8,10]
)

print("\nRFE")
print(summary_rfe)
```

```python
# Combine all results
summary = pd.concat(

[summary_k.rename(columns={"K_Value":"Feature_Count"}).assign(Method="SelectK"),

summary_pca.rename(columns={"PCA":"Feature_Count"}).assign(Method="PCA"),

summary_rfe.rename(columns={"RFE":"Feature_Count","Feature_Model":"Best_Model"}).assign(Method="RFE")],

ignore_index=True)

best = summary.loc[
summary["Best_Score"].idxmax()
]

print("\nFinal Best Result")
print(best)
```
