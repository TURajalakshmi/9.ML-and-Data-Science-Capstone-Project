```python
from statsmodels.stats.outliers_influence import variance_inflation_factor
def calc_vif(X):
    vif=pd.DataFrame()
    vif["variable"]=X.columns
    vif["VIF"]=[variance_inflation_factor(X.values, i) for i in range (X.shape[1])]
    return (vif)
```

### T-Test & Hypothesis test

```python
from scipy.stats import ttest_ind

Dataset = dataset.dropna(subset=['OverallQual', 'SalePrice'])

High_Quality = Dataset[Dataset['OverallQual'] >= 7]['SalePrice']
Low_Quality = Dataset[Dataset['OverallQual'] < 7]['SalePrice']

ttest_ind(High_Quality, Low_Quality)
```

```python
# Statement: Test whether the average SalePrice of high-quality houses and low-quality houses are the same.

# Null Hypothesis (H₀):There is no significant difference in the average SalePrice between high-quality houses (OverallQual ≥ 7)
# and low-quality houses (OverallQual < 7).

# Alternate Hypothesis (H₁): There is a significant difference in the average SalePrice between high-quality houses
# and low-quality houses.

# p-value = 7.77e-224 (p < 0.05)

# Decision: Reject the null hypothesis (H₀). Accept the alternate hypothesis (H₁).

# Conclusion: There is a statistically significant difference in average SalePrice between high-quality and low-quality houses.
```

```python
from scipy.stats import ttest_ind

Dataset = dataset.dropna(subset=['CentralAir', 'SalePrice'])

CentralAir_Yes = Dataset[Dataset['CentralAir'] == 'Y']["SalePrice"]
CentralAir_No = Dataset[Dataset['CentralAir'] == 'N']["SalePrice"]

ttest_ind(CentralAir_Yes, CentralAir_No)
```

```python
# Statement: Test whether the average SalePrice of houses with Central Air and houses without Central Air are the same.

# Null Hypothesis (H₀): There is no significant difference in average SalePrice between houses with Central Air and houses without Central Air.

# Alternate Hypothesis (H₁): There is a significant difference in average SalePrice between houses with Central Air and houses without Central Air.

# p-value = 2.73e-28 (p < 0.05)

# Decision: Reject the null hypothesis (H₀). Accept the alternate hypothesis (H₁).

# Conclusion: There is a statistically significant difference in average SalePrice between houses with Central Air and houses without Central Air.
```

```python
from scipy.stats import ttest_ind

Dataset = dataset.dropna(subset=['YearBuilt', 'SalePrice'])

New_House = Dataset[Dataset['YearBuilt'] >= 2000]['SalePrice']
Old_House = Dataset[Dataset['YearBuilt'] < 2000]['SalePrice']

ttest_ind(New_House, Old_House)
```

```python
# Statement: Test whether the average SalePrice of new houses and old houses are the same.

# Null Hypothesis (H₀): There is no significant difference in average SalePrice between houses built in or after 2000 and houses built before 2000.

# Alternate Hypothesis (H₁): There is a significant difference in average SalePrice between houses built in or after 2000 and houses built before 2000.

# p-value = 1.25e-93 (p < 0.05)

# Decision: Reject the null hypothesis (H₀). Accept the alternate hypothesis (H₁).

# Conclusion: There is a statistically significant difference in average SalePrice between new houses and old houses.
```

```python
from scipy.stats import ttest_ind

Dataset = dataset.dropna(subset=['GrLivArea', 'SalePrice'])

Large_House = Dataset[Dataset['GrLivArea'] >= Dataset['GrLivArea'].median()]['SalePrice']
Small_House = Dataset[Dataset['GrLivArea'] < Dataset['GrLivArea'].median()]['SalePrice']

ttest_ind(Large_House, Small_House)
```

```python
# Statement: Test whether the average SalePrice of large houses and small houses are the same.

# Null Hypothesis (H₀): There is no significant difference in average SalePrice between large houses
# (GrLivArea greater than or equal to median value) and small houses (GrLivArea less than median value).

# Alternate Hypothesis (H₁): There is a significant difference in average SalePrice between large houses
# (GrLivArea greater than or equal to median value) and small houses (GrLivArea less than median value).

# p-value = 3.59e-154 (p < 0.05)

# Decision: Reject the null hypothesis (H₀). Accept the alternate hypothesis (H₁).

# Conclusion: There is a statistically significant difference in average SalePrice between large houses and small houses.
```
